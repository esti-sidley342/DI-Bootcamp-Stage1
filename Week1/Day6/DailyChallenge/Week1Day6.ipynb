```python
# Exercise 1
# Ask the user to enter a word
word = input("Enter a word: ")

# Create an empty dictionary to store characters and their positions
char_indices = {}

# Loop through each character and its index in the word
for index, char in enumerate(word):

    # Check if the character already exists as a key in the dictionary
    if char in char_indices:

        # If it exists, append the current index to the existing list
        char_indices[char].append(index)

    else:
        # If it does not exist, create a new key
        # and store the current index inside a list
        char_indices[char] = [index]

# Print the resulting dictionary
print(char_indices)
```

```python
# Exercise 2
# Dictionary containing item names and their prices
items_purchase = {
    "Phone": "$999",
    "Speakers": "$300",
    "Laptop": "$5,000",
    "PC": "$1200",
    "Water": "$1",
    "Bread": "$3",
    "Apple": "$2",
    "Honey": "$3",
    "Bananas": "$2",
    "Pan": "$50"

}

# Amount of money available in the wallet
wallet = "$120"

# Clean and convert the wallet amount to an integer
wallet_amount = int(wallet.replace("$", "").replace(",", ""))

# Create an empty basket
basket = []

# Loop through each item
for item, price in items_purchase.items():

    # Clean and convert the item price
    item_price = int(price.replace("$", "").replace(",", ""))

    # Check if the item can be purchased
    if wallet_amount >= item_price:
        basket.append(item)
        wallet_amount -= item_price

# Display the result
if len(basket) == 0:
    print("Nothing")
else:
    print(sorted(basket))
```
