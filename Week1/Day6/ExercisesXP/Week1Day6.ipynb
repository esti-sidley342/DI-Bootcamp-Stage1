```python
# Create a list of dictionary keys
keys = ['Ten', 'Twenty', 'Thirty']

# Create a list of corresponding values
values = [10, 20, 30]

# Combine the keys and values into key-value pairs using zip()
# and convert the pairs into a dictionary
result = dict(zip(keys, values))

# Display the resulting dictionary
print(result)
```

```python
# Store family members' names and ages in a dictionary
family = {
    "rick": 43,
    "beth": 13,
    "morty": 5,
    "summer": 8
}

# Initialise the total ticket cost
total_cost = 0

# Loop through each family member and their age
for name, age in family.items():

    # Children under 3 get in free
    if age < 3:
        ticket_price = 0

    # Children aged 3 to 12 pay $10
    elif age <= 12:
        ticket_price = 10

    # Anyone over 12 pays $15
    else:
        ticket_price = 15

    # Add the ticket price to the running total
    total_cost += ticket_price

    # Print the ticket price for the current family member
    print(f"{name.title()} (Age {age}): ${ticket_price}")

# Print the total cost for the entire family
print(f"\nTotal ticket cost: ${total_cost}")
```

```python
# User input version
# Create an empty dictionary to store family members
family = {}

# Ask the user how many family members there are
num_members = int(input("How many family members are going to the movies? "))

# Collect names and ages from the user
for i in range(num_members):
    name = input(f"\nEnter name for family member {i + 1}: ")
    age = int(input(f"Enter age for {name}: "))

    # Add the family member to the dictionary
    family[name] = age

# Initialise the total ticket cost
total_cost = 0

# Loop through each family member to calculate ticket prices
for name, age in family.items():

    # Children under 3 enter for free
    if age < 3:
        ticket_price = 0

    # Children aged 3 to 12 pay $10
    elif age <= 12:
        ticket_price = 10

    # Anyone over 12 pays $15
    else:
        ticket_price = 15

    # Add ticket price to the total cost
    total_cost += ticket_price

    # Display the ticket price for each person
    print(f"{name.title()} (Age {age}): ${ticket_price}")

# Display the final total cost
print(f"\nTotal ticket cost: ${total_cost}")
```

```python
# Create a dictionary called 'brand' containing Zara's information
brand = {
    "name": "Zara",
    "creation_date": 1975,
    "creator_name": "Amancio Ortega Gaona",
    "type_of_clothes": ["men", "women", "children", "home"],
    "international_competitors": ["Gap", "H&M", "Benetton"],
    "number_stores": 7000,
    "major_color": {
        "France": "blue",
        "Spain": "red",
        "US": ["pink", "green"]
    }
}

# Change the value of 'number_stores' from 7000 to 2
brand["number_stores"] = 2

# Print a sentence describing Zara's clients using the 'type_of_clothes' key
print(
    f"Zara's clients include "
    f"{', '.join(brand['type_of_clothes'])} customers."
)

# Add a new key called 'country_creation' with the value 'Spain'
brand["country_creation"] = "Spain"

# Check if 'international_competitors' exists and add 'Desigual' to the list
if "international_competitors" in brand:
    brand["international_competitors"].append("Desigual")

# Delete the 'creation_date' key from the dictionary
del brand["creation_date"]

# Print the last item in the international competitors list
print("\nLast competitor:")
print(brand["international_competitors"][-1])

# Print the major colours in the US
print("\nMajor colours in the US:")
print(brand["major_color"]["US"])

# Print the total number of keys in the dictionary
print("\nNumber of keys in the dictionary:")
print(len(brand))

# Print all keys in the dictionary
print("\nDictionary keys:")
print(brand.keys())
```

```python
# Create a list of Disney characters
users = ["Mickey", "Minnie", "Donald", "Ariel", "Pluto"]

# Dictionary 1: Map each character to its index

# Use enumerate() to get both the index and character
char_to_index = {}

for index, character in enumerate(users):
    char_to_index[character] = index

# Display the dictionary
print("Dictionary 1: Character -> Index")
print(char_to_index)

# Dictionary 2: Map each index to its character

# Use enumerate() to create index-character pairs
index_to_char = {}

for index, character in enumerate(users):
    index_to_char[index] = character

# Display the dictionary
print("\nDictionary 2: Index -> Character")
print(index_to_char)

# Dictionary 3: Sort characters alphabetically and map them to their new indices

# Sort the list alphabetically
sorted_users = sorted(users)

# Create a dictionary using the sorted list
sorted_char_to_index = {}

for index, character in enumerate(sorted_users):
    sorted_char_to_index[character] = index

# Display the dictionary
print("\nDictionary 3: Alphabetical Character -> Index")
print(sorted_char_to_index)
```
