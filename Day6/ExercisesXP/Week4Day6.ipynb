```python
import numpy as np
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt

# Exercise 1 : Matrix Operations
# Create a 3x3 matrix
matrix = np.array([
    [2, 1, 3],
    [1, 0, 2],
    [4, 1, 8]
])

# Display the original matrix
print("Matrix:")
print(matrix)

# Calculate the determinant of the matrix
# The determinant tells us whether the matrix is invertible.
# If the determinant is 0, the matrix does not have an inverse.
determinant = np.linalg.det(matrix)

# Display the determinant
print("\nDeterminant:")
print(determinant)

# Check whether the matrix can be inverted
if determinant != 0:

    # Calculate the inverse of the matrix
    inverse_matrix = np.linalg.inv(matrix)

    # Display the inverse matrix
    print("\nInverse Matrix:")
    print(inverse_matrix)

else:
    # Inform the user that the matrix is singular and therefore has no inverse
    print("\nThe matrix has no inverse because its determinant is 0.")
```

```python
# Exercise 2 : Statistical Analysis
# Generate an array containing 50 random numbers.
# np.random.rand(50) creates 50 random floating-point numbers between 0 and 1.
random_numbers = np.random.randint(1, 101, size=50)

# Display the generated array so we can see the values.
print("Random Numbers:")
print(random_numbers)

# Calculate the mean (average) of the array.
# The mean is the sum of all values divided by the total number of values.
mean_value = np.mean(random_numbers)

# Calculate the median of the array.
# The median is the middle value when the numbers are sorted.
median_value = np.median(random_numbers)

# Calculate the standard deviation of the array.
# Standard deviation measures how spread out the values are from the mean.
# Calculate and round the standard deviation to 2 decimal places.
std_deviation = round(np.std(random_numbers), 2)

# Display the calculated mean.
print("\nMean:")
print(mean_value)

# Display the calculated median.
print("\nMedian:")
print(median_value)

# Display the calculated standard deviation.
print("\nStandard Deviation:")
print(std_deviation)
```

```python
# Exercise 3 : Date Manipulation

# Create an array of dates for every day in January 2023.
# The end date is excluded, so 2023-02-01 is used to include all of January.
dates = np.arange('2023-01-01', '2023-02-01', dtype='datetime64[D]')

# Display the original dates in YYYY-MM-DD format.
print("Original Dates:")
print(dates)

# Convert each date to a string and replace "-" with "/". This changes the format from YYYY-MM-DD to YYYY/MM/DD.
formatted_dates = np.array([
    str(date).replace('-', '/')
    for date in dates
])

# Display the dates in the new format.
print("\nDates in YYYY/MM/DD Format:")
print(formatted_dates)
```

```python
# Exercise 4 : Data Manipulation with NumPy and Pandas

# Set a random seed so the results are reproducible.
# This means the same random numbers will be generated each time.
np.random.seed(42)

# Create a DataFrame with 10 rows and 3 columns of random integers. The integers will range from 1 to 100.
df = pd.DataFrame({
    'A': np.random.randint(1, 101, 10),
    'B': np.random.randint(1, 101, 10),
    'C': np.random.randint(1, 101, 10)
})

# Display the original DataFrame.
print("Original DataFrame:")
print(df)

# Conditional Selection of Data

# Select only the rows where the value in column A is greater than 50.
filtered_df = df[df['A'] > 50]

# Display the filtered DataFrame.
print("\nRows where column A is greater than 50:")
print(filtered_df)

# Aggregation Functions

# Calculate the sum of each column.
column_sums = df.sum()

# Display the sum of each column.
print("\nSum of each column:")
print(column_sums)

# Calculate the average (mean) of each column.
column_means = df.mean()

# Display the average of each column.
print("\nAverage of each column:")
print(column_means)

# Count the number of non-null values in each column.
print(df.count())

# Find the median value in each column.
print(df.median())

# Calculate the standard deviation of each column.
print(df.std())

# Calculate the variance of each column.
print(df.var())
```

Exercise 5 : Image Representation

In NumPy, an image is represented as an array of numbers where each number corresponds to a pixel value.

Grayscale image: A 2D array (rows × columns).

Each element represents the brightness of a pixel.
Values typically range from:
0 = black
255 = white
Values in between = shades of gray

Colour image (RGB): A 3D array (rows × columns × 3).

The third dimension stores the Red, Green, and Blue values for each pixel.

```python
# Open the image from the current folder.
image = Image.open("Profilepic.png")

# Convert the image to grayscale.
# 'L' mode represents an 8-bit grayscale image.
grayscale_image = image.convert("L")

# Convert the grayscale image into a NumPy array.
# Each element in the array represents a pixel intensity:
# 0 = black, 255 = white.
image_array = np.array(grayscale_image)

# Display the NumPy array.
print("Grayscale Image Array:")
print(image_array)

# Display the dimensions of the image.
# This shows the number of rows and columns (height, width).
print("\nImage Shape:")
print(image_array.shape)

# Save the grayscale image to a new file.
grayscale_image.save("Profilepic_Grayscale.png")

# Inform the user that the conversion is complete.
print("\nGrayscale image saved as 'Profilepic_Grayscale.png'")
```

```python
# Exercise 6 : Basic Hypothesis Testing

# Set a random seed so that the results are reproducible.  The same random numbers will be generated each time the code runs.
np.random.seed(42)

# Productivity scores of employees before the training programme. 
# Scores are generated from a normal distribution with mean = 50 and standard deviation = 10.
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivity scores after the training programme.
# Each employee receives an average improvement of 5 points, with some random variation.
productivity_after = productivity_before + np.random.normal(
    loc=5,
    scale=3,
    size=30
)

# Calculate the average productivity before training.
mean_before = np.mean(productivity_before)

# Calculate the average productivity after training.
mean_after = np.mean(productivity_after)

# Calculate the productivity improvement for each employee.
improvement = productivity_after - productivity_before

# Calculate the average improvement across all employees.
mean_improvement = np.mean(improvement)

# Display the results.
print("Average Productivity Before Training:", round(mean_before, 2))
print("Average Productivity After Training:", round(mean_after, 2))
print("Average Improvement:", round(mean_improvement, 2))

# Test the hypothesis.
# If the mean productivity after training is greater than before,
# there is evidence that the programme improved productivity.
if mean_after > mean_before:
    print("\nConclusion:")
    print("The results support the hypothesis that the training programme improved productivity.")
else:
    print("\nConclusion:")
    print("The results do not support the hypothesis that the training programme improved productivity.")
```

```python
# Exercise 7 : Complex Array Comparison

# Create the first array.
array1 = np.array([10, 25, 30, 45, 50])

# Create the second array.
array2 = np.array([5, 30, 20, 45, 60])

# Compare the arrays element by element.
# The result will be True where the element in array1 is greater than the corresponding element in array2 and False otherwise.
comparison = array1 > array2

# Display both arrays.
print("First Array:", array1)
print("Second Array:", array2)

# Display the boolean comparison result.
print("\nElements in the first array greater than the second:")
print(comparison)

# Create a DataFrame to display the results as a table.
results_table = pd.DataFrame({
    "Array 1": array1,
    "Array 2": array2,
    "Array 1 > Array 2": comparison
})

# Display the table.
print(results_table)
```

```python
# Exercise 8 : Time Series Data Manipulation

# Generate a date range for the entire year 2023. 'D' specifies daily frequency.
dates = pd.date_range(start='2023-01-01', end='2023-12-31', freq='D')

# Generate random values for each date. These represent sales, temperature, stock prices, etc.
values = np.random.randint(1, 101, len(dates))

# Create a DataFrame with dates as the index.
time_series = pd.DataFrame({
    'Value': values
}, index=dates)

# Display the first few rows of the dataset.
print("Time Series Data:")
print(time_series.head())

# January to March (Q1)

# Slice all data from January through March.
jan_to_mar = time_series['2023-01-01':'2023-03-31']

print("\nJanuary to March:")
print(jan_to_mar)

# April to June (Q2)

# Slice all data from April through June.
apr_to_jun = time_series['2023-04-01':'2023-06-30']

print("\nApril to June:")
print(apr_to_jun)

# July to September (Q3)

# Slice all data from July through September.
jul_to_sep = time_series['2023-07-01':'2023-09-30']

print("\nJuly to September:")
print(jul_to_sep)

# October to December (Q4)

# Slice all data from October through December.
oct_to_dec = time_series['2023-10-01':'2023-12-31']

print("\nOctober to December:")
print(oct_to_dec)
```

```python
# Exercise 9 : Data Conversion

# NumPy Array to Pandas DataFrame
# Create a 2D NumPy array.
numpy_array = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# Display the original NumPy array.
print("Original NumPy Array:")
print(numpy_array)

# Convert the NumPy array to a Pandas DataFrame. The columns parameter assigns names to each column.
df = pd.DataFrame(
    numpy_array,
    columns=["A", "B", "C"]
)

# Display the DataFrame.
print("\nPandas DataFrame:")
print(df)

# Pandas DataFrame to NumPy Array
# Convert the DataFrame back to a NumPy array to_numpy() returns the underlying NumPy representation.
converted_array = df.to_numpy()

# Display the converted NumPy array.
print("\nConverted NumPy Array:")
print(converted_array)
```

```python
# Exercise 10 : Basic Visualization

# Generate 20 random integers between 1 and 100. These will be the values plotted on the y-axis.
data = np.random.randint(1, 101, size=20)

# Create x-axis values from 1 to 20. These represent the position of each data point.
x = np.arange(1, 21)

# Create a new figure for the plot.
plt.figure(figsize=(8, 4))

# Plot the data as a line graph. marker='o' displays a circle at each data point.
plt.plot(x, data, marker='o')

# Add a title to the graph.
plt.title("Line Graph of Random Numbers")

# Label the x-axis.
plt.xlabel("Data Point")

# Label the y-axis.
plt.ylabel("Value")

# Add grid lines to make the graph easier to read.
plt.grid(True)

# Display the graph.
plt.show()
```
