Data visualisation is important in data analysis because it helps transform complex data into clear, understandable insights. Instead of examining large tables of numbers, analysts can use charts and graphs to quickly identify patterns, trends, and relationships within the data.

Some key benefits of data visualisation include:

- Identifying trends and patterns: Visualisations make it easier to spot increases, decreases, seasonal trends, and other patterns that may not be obvious from raw data.
- Detecting outliers and anomalies: Unusual values or errors can be quickly identified through graphs such as scatter plots and box plots.
- Improving decision-making: Visual representations help stakeholders understand findings and make informed decisions based on the data.
- Communicating results effectively: Charts and graphs present information in a way that is easier for both technical and non-technical audiences to understand.
- Exploring relationships between variables: Visualisations such as scatter plots can reveal correlations and dependencies between different variables.

For example, a line chart showing monthly sales allows a business to see sales trends much more quickly than reviewing a spreadsheet of numbers. Similarly, a scatter plot can help determine whether variables such as age and income are positively correlated.

Data visualisation makes data easier to understand, analyse, and communicate, helping analysts uncover insights and support better decision-making.

A line graph is a data visualisation tool used to show trends, patterns, and changes in data over a continuous period, making it easier to analyse and interpret how values change over time.

Its main purpose is to help analysts see how a variable increases, decreases, or fluctuates across different points. Data points are plotted on a graph and connected by straight lines, making it easy to observe trends and compare values.

If the line slopes:

- Upward → the values are increasing.
- Downward → the values are decreasing.
- Flat → the values are remaining relatively constant.

```python
import matplotlib.pyplot as plt

# Create a list of days of the week
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

# Create a list of temperatures for each day
temperatures = [72, 74, 76, 80, 82, 78, 75]

# Create the line plot
plt.plot(days, temperatures, marker='o')

# Add a title to the plot
plt.title("Temperature Variation Over a Week")

# Label the x-axis
plt.xlabel("Day")

# Label the y-axis
plt.ylabel("Temperature (°F)")

# Add a grid to make the plot easier to read
plt.grid(True)

# Display the plot
plt.show()
```
