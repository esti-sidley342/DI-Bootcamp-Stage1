```python
from scipy import stats
data = [12, 15, 13, 12, 18, 20, 22, 21]

# Calculate the mean
mean = stats.tmean(data)

# Calculate the median
median = stats.median_abs_deviation([0])  # placeholder to show SciPy import
median = stats.scoreatpercentile(data, 50)

# Calculate the variance
variance = stats.tvar(data)

# Calculate the standard deviation
std_dev = variance ** 0.5

# Display the results

print("Dataset:", data)
print("Mean:", mean)
print("Median:", median)
print("Sample Variance:", variance)
print("Sample Standard Deviation:", std_dev)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Define the distribution parameters

# Mean (centre of the distribution)
mean = 50

# Standard deviation (spread of the distribution)
std_dev = 10

# Generate x-values
# Create 100 evenly spaced values between 20 and 80 to cover most of the distribution
x = np.linspace(20, 80, 100)

# Calculate the probability density
# norm.pdf() computes the probability density function (PDF) for each x-value
y = norm.pdf(x, mean, std_dev)

# Plot the distribution
plt.plot(x, y, label="Normal Distribution")

# Add a title to the graph
plt.title("Normal Distribution (Mean = 50, Std Dev = 10)")

# Label the x-axis
plt.xlabel("Values")

# Label the y-axis
plt.ylabel("Probability Density")

# Display a legend
plt.legend()

# Add grid lines for easier reading
plt.grid(True)

# Show the plot
plt.show()
```

```python
import numpy as np
from scipy.stats import ttest_ind

data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

#Independent t-test
t_statistic, p_value = ttest_ind(data1, data2)

# Display the results

print("T-Statistic:", t_statistic)
print("P-Value:", p_value)

# Interpret the Results

# Common significance level (alpha)
alpha = 0.05

if p_value < alpha:
    print("The difference between the two groups is statistically significant.")
else:
    print("There is no statistically significant difference between the two groups.")
```

```python
from scipy.stats import linregress

# House sizes (independent variable, x)
house_sizes = [50, 70, 80, 100, 120]

# House prices (dependent variable, y)
house_prices = [150000, 200000, 210000, 250000, 280000]

# Perform Linear Regression
# linregress() calculates:
# - slope
# - intercept
# - correlation coefficient (r)
# - p-value
# - standard error
result = linregress(house_sizes, house_prices)

# Extract the slope and intercept
slope = result.slope
intercept = result.intercept

# Predict the price of a 90 m² house
house_size = 90

# Formula of a regression line:
# y = mx + b
# where:
# m = slope
# b = intercept
predicted_price = slope * house_size + intercept

# Display Results
print("Slope:", slope)
print("Intercept:", intercept)
print(f"Predicted Price for 90 m²: ${predicted_price:,.0f}")
print("The slope (1,845.24) represents the expected change in house price for every additional square metre of house size. For every extra 1 square metre of living space, the house price is expected to increase by approximately 1,845 currency units. This positive slope indicates a positive relationship between house size and house price: larger houses tend to be more expensive.")
```

```python
from scipy.stats import f_oneway

fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

# Perform a one-way ANOVA test
f_value, p_value = f_oneway(
    fertilizer_1,
    fertilizer_2,
    fertilizer_3
)

# Display the results
print(f"F-value: {f_value:.2f}")
print(f"P-value: {p_value:.4f}")
print("If p is less than 0.05, we reject the null hypothesis. This means there is strong statistical evidence that at least one fertilizer produces a different average plant growth than the others.")
print(
    "In practical terms:\n"
    "\n"
    "- Fertilizer 1 average growth = 5.8 cm\n"
    "- Fertilizer 2 average growth = 7.8 cm\n"
    "- Fertilizer 3 average growth = 4.0 cm\n"
    "\n"
    "The differences are large enough that they are unlikely to have occurred by random chance alone."
)

print("As p is greater than 0.05, we would fail to reject the null hypothesis. This means: We would not have enough evidence to conclude that the fertilizers differ in effectiveness.")
print("Any differences observed in the sample means could reasonably be due to random variation.")
print("We would treat the fertilizers as having similar effects on plant growth, based on the available data")
```

```python
from scipy.stats import binom

# Number of coin flips
n = 10

# Probability of getting a head
p = 0.5

# Number of heads we're interested in
k = 5

# Calculate the probability of getting exactly 5 heads
probability = binom.pmf(k, n, p)

# Display the result
print(f"Probability of getting exactly 5 heads: {probability:.4f}")
```

```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr

# Create the dataset
data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})

# Extract the variables
age = data['age']
income = data['income']

# Pearson Correlation
# Measures the strength and direction of a linear relationship between two variables.
pearson_corr, pearson_p = pearsonr(age, income)

# Spearman Correlation
# Measures the strength and direction of a monotonic relationship based on ranks.
spearman_corr, spearman_p = spearmanr(age, income)

# Display Results
print(f"Pearson Correlation Coefficient: {pearson_corr:.4f}")
print(f"Pearson P-Value: {pearson_p:.4f}")

print(f"\nSpearman Correlation Coefficient: {spearman_corr:.4f}")
print(f"Spearman P-Value: {spearman_p:.4f}")
```
