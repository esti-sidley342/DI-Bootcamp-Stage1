Exercise 1:
Data analysis is the systematic process of inspecting, cleansing, transforming and modelling raw data to discover useful information, come to conclusions on past events and why they happened, and to inform better decision making.
There are four types of data analysis:
- descriptive - which describes what happened, usually using metrics such as totals, averages, percentage changes etc. An example of this could be a sales report.
- diagnostic - investigates root causes and/or anomalies and looks at why events happened
- predictive - uses historical data to model what could happen based on past events or behaviours
- prescriptive - advises what action could be taken based on the predictive analysis.

Raw data itself is not useful until it is processed. To make it useful, the following has to be happen:
- Inspecting - understand it shape, types, distribution and quality. I.e. checking for data types such as if it is text, numbers, dates. Spotting whether the data is complete or if it's missing some values.
- Cleaning - removing duplicate data, or standardizing (making sure that Form and form are the same and there isn't the mismatch by upper/lower case), handling missing data by filling in averages or a placeholder, or standardizing date formats to make sure they are all the same.
- Transformation - make sure the data matches the specific needs of for the analysis required, such as giving ranges for numbers and categorising (i.e. red for bad, amber for warning, green for good). Metrics can also combined and aggregated to calculate monthly averages, for example, or create new predictive variables from the existing ones.
- Modelling - At this point, machine learning can be used to prepare predictions on a data set and run predictions based on the available historical data, looking at patterns,

Data analysis is essential in the modern world. Raw data is generated constantly and is huge, but is meaningless unless it is processed and interpreted. For example - data is generated every time a user clicks on a link on social media, but it doesn't become useful until a pattern of behaviour is established to see what they are clicking on in order to keep providing that content to keep them interested. If they are bombarded with links which are not of interest to them, they may stop using that platform. Therefore it is essential to record and analyse what they have clicked on, whether they stayed long enough on a link (because they may have clicked it by accident - therefore if they leave after only a few seconds then it's not really worth noting). If their likes and interests are noted, then predictions can be made on what will keep them engaged, and prescriptive action can be taken to keep providing them with links that are linked to their interest.
Data analysis means that informed decision making can be made, instead of pure guess work, and it is quantifiable. It reduces costs and increases efficiency so that resources are not wasted on products or advertisements which do not engage people - or in the case of health care, where medicines are shown to be not as efficient, then less investment is made in them.
Deeper customer insights can be recorded with feedback, purchase histories etc which will help businesses to anticipate what customers are after.
It helps with mitigating risk and making more accurate predictions through all the four types of data analysis, as organisations can then spot market trends to forecast outcomes, or proactively address risks without losing huge quantities of revenue.
This isn't just linked to impacts on marketing and healthcare - this can also influence public policy (such as why did a particular political party lose an election - which policy cost them votes?), or education where learning outcomes translate into better career prospects.

Three key areas where data analysis is used today:
 - healthcare - used for disease forecasting and to support clinical decisions, patient monitoring tracking disease outcomes. Data analysis is always instrumental when it comes to tracking the latest disease outbreaks in order to place resources where they will be best utilised. For example: when coronavirus was first detected, patterns of where it was emerging, how infectious it is, fatality rates etc had to be analysed in order to influence policy on whether it was worth shutting down education, international travel, etc.
- banking and finance - data analysis is used to detect fraud (i.e. whether a transaction is completely out of character for a customer, in that case it can initially be blocked until the customer confirms that it is them or not initiating a transaction. It is also used for credit scoring to see whether a customer is more likely to pay back money they borrowed.
- retail - used to see where demand is and whether it is worth increasing the inventory of specific products and when, and to personalise customer recommendations based on previous purchasing habits.

### Extracting a ZIP File
You can use Python's built-in `zipfile` module to extract files directly within Colab.

### Uploading Files from Your Local Machine
You can use `google.colab.files.upload()` to upload your ZIP file directly from your computer to the notebook workspace.

```python
#Exercise 2
import zipfile
import os

# List of ZIP files to extract
zip_files = [
    '/content/Credit Card Approvals (Clean Data).zip',
    '/content/Global Trends in Mental Health Disorder.zip',
    '/content/How Much Sleep Do Americans.zip'
]

# Base directory for extraction
base_extraction_path = '/content/extracted_data'

for zip_file_path in zip_files:
    if os.path.exists(zip_file_path):
        # Create a specific folder name for each zip file to avoid conflicts
        folder_name = os.path.splitext(os.path.basename(zip_file_path))[0]
        extraction_folder = os.path.join(base_extraction_path, folder_name)

        os.makedirs(extraction_folder, exist_ok=True)

        with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
            zip_ref.extractall(extraction_folder)

        print(f"Extracted: {os.path.basename(zip_file_path)}")
        print(f"Location: {extraction_folder}")
        print("Contents:", os.listdir(extraction_folder))
        print("-" * 50)
    else:
        print(f"File not found: {zip_file_path}")
```

```python
#Exercise 3
import pandas as pd

#This is a very simple exercise - if the data is an integer or a float, it is quantitative. This includes data on age, years employed, debt owed, amount of sleep in hours etc. For the mental health disorders, this is included because the different types are recorded as percentages. For anything else, it will be qualitative.

# Define the paths for the three datasets
datasets = {
    'Credit Card Approvals (Clean Data)': '/content/extracted_data/Credit Card Approvals (Clean Data)/Credit Card Approvals (Clean Data)/clean_dataset.csv',
    'Global Trends in Mental Health Disorder': '/content/extracted_data/Global Trends in Mental Health Disorder/Global Trends in Mental Health Disorder/Mental health Depression disorder Data.csv',
    'How Much Sleep Do Americans': '/content/extracted_data/How Much Sleep Do Americans/How Much Sleep Do Americans/Time Americans Spend Sleeping.csv'
}

for name, path in datasets.items():
    print(f"=== {name} ===")
    # Read data (using low_memory=False for mental health dataset just in case)
    df_temp = pd.read_csv(path, low_memory=False)

    quantitative_cols = []
    qualitative_cols = []

    for col in df_temp.columns:
        # Check if the column is numeric (integer or float)
        if pd.api.types.is_numeric_dtype(df_temp[col]):
            # We exclude binary indicators (0/1) or year/index columns sometimes, but statistically they are numeric
            quantitative_cols.append(col)
        else:
            qualitative_cols.append(col)

    print("📊 Quantitative (Numerical) Columns:")
    for col in quantitative_cols:
        print(f"  - {col} ({df_temp[col].dtype})")

    print("\n🏷️ Qualitative (Categorical) Columns:")
    for col in qualitative_cols:
        print(f"  - {col} ({df_temp[col].dtype})")
    print("\n" + "="*50 + "\n")
```

### Displaying First Three Rows of Each Dataset
Let's locate the extracted data files (like CSVs or text files) inside each folder, load them, and inspect their first 3 rows.

```python
import os
import pandas as pd

base_path = '/content/extracted_data'
datasets = {}

# Walk through the extracted folders to locate files
for root, dirs, files in os.walk(base_path):
    for file in files:
        if file.endswith('.csv') or file.endswith('.txt') or file.endswith('.tsv'):
            file_path = os.path.join(root, file)
            parent_folder = os.path.basename(os.path.dirname(root))
            if not parent_folder or parent_folder == 'extracted_data':
                parent_folder = os.path.basename(root)
            datasets[parent_folder] = file_path

# Load and display the first three rows of each detected dataset
for dataset_name, file_path in datasets.items():
    print(f"=== Dataset: {dataset_name} ===")
    print(f"File path: {file_path}")
    try:
        # Load the dataset using pandas
        df = pd.read_csv(file_path)
        print(f"Shape: {df.shape[0]} rows, {df.shape[1]} columns\n")
        print("First 3 rows:")
        display(df.head(3))
    except Exception as e:
        print(f"Could not read file: {e}")
    print("\n" + "="*60 + "\n")
```
